```python
from neuroCombat import neuroCombat
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('/Users/charlie/Desktop/my_projects/neurotransmitter/github/data/mind_node.csv')
data = data.values

covariates = pd.read_csv('/Users/charlie/Desktop/my_projects/neurotransmitter/github/data/demos_long.csv')
# Remove the first column
covariates = covariates.drop(covariates.columns[3], axis=1)
covariates = covariates.drop(covariates.columns[0], axis=1)
covariates = covariates.drop(covariates.columns[4], axis=1)
print(covariates)

#Specify categorical columns
categorical_cols = ['sex', 'group']
#Specify scanner variable
##Here I assigned a different value to each scanner in each site in each study
batch_col = 'site'

data_combat = neuroCombat(dat=data,
    covars=covariates,
    batch_col=batch_col,
    categorical_cols=categorical_cols)["data"]

file_path = '/Users/charlie/Desktop/my_projects/neurotransmitter/github/results/data_combat.csv'
np.savetxt(file_path, data_combat, delimiter=',', fmt='%s')

print(data)
print(data_combat)
```
